```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import convolve
```

```python
img=cv2.imread("/content/image11.jfif",cv2.IMREAD_GRAYSCALE)
```

```python
plt.figure(figsize=(6,4))
plt.imshow(img, cmap='gray')
plt.show()
```

```python
# sharpen filter
#if imge is brighter compared to the background then need to supress baaccckground that is why -1 and 5
bright_shrpen=[[0,-1,0],[-1,5,-1],[0,-1,0]]

#if image is darker compared to the background then need to enhance background that is why 1 and -5
dark_shrpen=[[0,1,0],[1,-5,1],[0,1,0]]
```

```python
# blur kernel
# this are the eqully divided meenas every pixel eqaully treated
blur7=np.ones((7,7),dtype=float)/49
blur3=np.ones((3,3),dtype=float)/9
blur5=np.ones((5,5),dtype=float)/25

#they are not equally divided means some pixels are given more weightage than others
#so only some areas are blurred and some areas are not blurred
blur5n=np.ones((5,5),dtype=float)/2
blur7n=np.ones((7,7),dtype=float)/2
```

```python
kernel=blur5n
im=np.array(img,dtype=float)
img2=convolve(im,kernel)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2,cmap='gray')
```
