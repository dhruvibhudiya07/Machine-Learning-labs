```python
# Q5: Ratio of Duplicate Questions Asked in Each Month
# Period: 2008 to 2018

import pandas as pd
import matplotlib.pyplot as plt

# Convert question creation date
questions["creation_date"] = pd.to_datetime(
    questions["creation_date"],
    errors="coerce"
)

# Make sure duplicate links are LinkTypeId = 3
duplicate_links = postlinks[
    postlinks["link_type_id"] == 3
].copy()

# IMPORTANT:
# post_id = duplicate question
duplicate_question_ids = set(
    duplicate_links["post_id"].dropna()
)

# Mark duplicate questions
questions["is_duplicate"] = questions["id"].isin(
    duplicate_question_ids
)

# Select questions from 2008 to 2018
q = questions[
    (questions["creation_date"] >= "2008-01-01") &
    (questions["creation_date"] < "2019-01-01")
].copy()

# Create month
q["month"] = q["creation_date"].dt.to_period("M")

# Total questions in each month
monthly_total = q.groupby("month").size()

# Duplicate questions in each month
monthly_duplicate = (
    q[q["is_duplicate"]]
    .groupby("month")
    .size()
)

# Combine both
monthly_data = pd.DataFrame({
    "Total Questions": monthly_total,
    "Duplicate Questions": monthly_duplicate
}).fillna(0)

# Calculate percentage
monthly_data["Duplicate Ratio (%)"] = (
    monthly_data["Duplicate Questions"] /
    monthly_data["Total Questions"] * 100
)

# Display result
display(monthly_data.head(15))

# -------------------------
# LINE GRAPH
# -------------------------

plt.figure(figsize=(16, 7))

plt.plot(
    monthly_data.index.astype(str),
    monthly_data["Duplicate Ratio (%)"],
    marker="o",
    linewidth=1
)

plt.xlabel("Month")
plt.ylabel("Duplicate Questions (%)")
plt.title("Monthly Ratio of Duplicate Questions (2008–2018)")

plt.xticks(
    range(0, len(monthly_data), 6),
    monthly_data.index.astype(str)[::6],
    rotation=45
)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Q6 - Improved meaningful analysis

import re
import pandas as pd
import matplotlib.pyplot as plt

# Duplicate question IDs
duplicate_question_ids = set(
    postlinks.loc[
        postlinks["link_type_id"] == 3,
        "post_id"
    ].dropna()
)

# Count questions and duplicate questions for each tag
tag_total = {}
tag_duplicate = {}

for _, row in questions.iterrows():

    if pd.isna(row["tags"]):
        continue

    tags = re.findall(r"<([^>]+)>", row["tags"])

    for tag in set(tags):

        tag_total[tag] = tag_total.get(tag, 0) + 1

        if row["id"] in duplicate_question_ids:
            tag_duplicate[tag] = tag_duplicate.get(tag, 0) + 1

# Create result table
result = []

for tag, total in tag_total.items():

    duplicate = tag_duplicate.get(tag, 0)

    percentage = (duplicate / total) * 100

    result.append([
        tag,
        total,
        duplicate,
        percentage
    ])

tag_result = pd.DataFrame(
    result,
    columns=[
        "Tag",
        "Total Questions",
        "Duplicate Questions",
        "Duplicate Percentage"
    ]
)

# Remove very rare tags
# Only consider tags appearing in at least 100 questions
meaningful_tags = tag_result[
    tag_result["Total Questions"] >= 100
].copy()

# Top 20 based on duplicate percentage
top20 = meaningful_tags.sort_values(
    "Duplicate Percentage",
    ascending=False
).head(20)

print("Top 20 Tags Based on Duplicate Question Percentage")
display(top20)

# Bar chart
plt.figure(figsize=(14, 7))

plt.bar(
    top20["Tag"],
    top20["Duplicate Percentage"]
)

plt.xlabel("Tag")
plt.ylabel("Duplicate Questions (%)")
plt.title("Top 20 Tags Based on Duplicate Question Percentage")

plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

```python
# Q7: Time Required to Close Duplicate Questions

import pandas as pd
import matplotlib.pyplot as plt

# Convert dates
questions["creation_date"] = pd.to_datetime(
    questions["creation_date"],
    errors="coerce"
)

post_history["creation_date"] = pd.to_datetime(
    post_history["creation_date"],
    errors="coerce"
)

# -----------------------------------------
# 1. Find duplicate questions
# -----------------------------------------

duplicate_question_ids = set(
    postlinks.loc[
        postlinks["link_type_id"] == 3,
        "post_id"
    ].dropna()
)

# -----------------------------------------
# 2. Find closing events
# ph_type_id = 10 → Post Closed
# comment = 1 → Exact Duplicate (old Stack Exchange data)
# -----------------------------------------

close_events = post_history[
    (post_history["ph_type_id"] == 10) &
    (post_history["comment"].astype(str) == "1")
].copy()

# Keep only duplicate questions
close_events = close_events[
    close_events["post_id"].isin(duplicate_question_ids)
]

# -----------------------------------------
# 3. Get question creation date
# -----------------------------------------

creation_dates = questions[
    questions["id"].isin(duplicate_question_ids)
][["id", "creation_date"]].copy()

creation_dates = creation_dates.rename(
    columns={"id": "post_id", "creation_date": "question_creation"}
)

# -----------------------------------------
# 4. Merge creation and closing dates
# -----------------------------------------

q7 = close_events.merge(
    creation_dates,
    on="post_id",
    how="inner"
)

# -----------------------------------------
# 5. Calculate time required
# -----------------------------------------

q7["time_to_close_hours"] = (
    q7["creation_date"] - q7["question_creation"]
).dt.total_seconds() / 3600

# Remove invalid/negative values
q7 = q7[
    q7["time_to_close_hours"] >= 0
]

# -----------------------------------------
# 6. Summary
# -----------------------------------------

print("Number of duplicate questions closed:", len(q7))
print(
    "Average time to close:",
    round(q7["time_to_close_hours"].mean(), 2),
    "hours"
)

print(
    "Median time to close:",
    round(q7["time_to_close_hours"].median(), 2),
    "hours"
)

display(
    q7[
        ["post_id", "question_creation",
         "creation_date", "time_to_close_hours"]
    ].head(10)
)

# -----------------------------------------
# 7. Bar chart
# -----------------------------------------

# Create time ranges
bins = [0, 1, 6, 12, 24, 72, 168, float("inf")]

labels = [
    "<1 hour",
    "1–6 hours",
    "6–12 hours",
    "12–24 hours",
    "1–3 days",
    "3–7 days",
    ">7 days"
]

q7["close_time_category"] = pd.cut(
    q7["time_to_close_hours"],
    bins=bins,
    labels=labels,
    right=True
)

distribution = (
    q7["close_time_category"]
    .value_counts()
    .reindex(labels)
    .fillna(0)
)

plt.figure(figsize=(10, 6))

plt.bar(
    distribution.index.astype(str),
    distribution.values
)

plt.xlabel("Time Required to Close")
plt.ylabel("Number of Duplicate Questions")
plt.title("Time Required to Close Duplicate Questions")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Q8: Reputation Distribution of Users Whose Questions Are Duplicates

import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------------------
# 1. Select duplicate questions
# -----------------------------------------

duplicate_questions = questions[
    questions["id"].isin(duplicate_question_ids)
].copy()

# -----------------------------------------
# 2. Add user reputation
# -----------------------------------------

duplicate_questions = duplicate_questions.merge(
    users[["id", "reputation"]],
    left_on="owner_user_id",
    right_on="id",
    how="left"
)

# Remove users whose reputation is missing
duplicate_questions = duplicate_questions.dropna(
    subset=["reputation"]
)

# -----------------------------------------
# 3. Create reputation categories
# -----------------------------------------

bins = [-1, 100, 500, 1000, 5000, 10000, 50000, float("inf")]

labels = [
    "0–100",
    "101–500",
    "501–1,000",
    "1,001–5,000",
    "5,001–10,000",
    "10,001–50,000",
    "50,000+"
]

duplicate_questions["reputation_category"] = pd.cut(
    duplicate_questions["reputation"],
    bins=bins,
    labels=labels
)

# -----------------------------------------
# 4. Count users in each category
# -----------------------------------------

reputation_distribution = (
    duplicate_questions
    .groupby(
        "reputation_category",
        observed=False
    )["owner_user_id"]
    .nunique()
    .reindex(labels)
    .fillna(0)
)

print("Number of users by reputation category:")
display(reputation_distribution)

# -----------------------------------------
# 5. Bar chart
# -----------------------------------------

plt.figure(figsize=(11, 6))

plt.bar(
    reputation_distribution.index.astype(str),
    reputation_distribution.values
)

plt.xlabel("User Reputation Range")
plt.ylabel("Number of Users")
plt.title(
    "Reputation Distribution of Users Whose Questions Are Closed as Duplicates"
)

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Q10: Percentage of Questions Covered by First 5000 Tags

import pandas as pd
import re
from collections import Counter

# Exact path of posts_short.csv
questions_path = "/kaggle/working/dataset/Arjun_Assignment_data/dataset-20210607T020316Z-001/dataset/posts_short.csv"

# Load questions
questions = pd.read_csv(questions_path)

print("Total Questions:", len(questions))

# -----------------------------------------
# Count all tags
# -----------------------------------------

tag_counter = Counter()

for tags in questions["tags"].fillna(""):
    
    found_tags = re.findall(
        r"<([^>]+)>",
        str(tags)
    )
    
    tag_counter.update(found_tags)

print("Total Unique Tags:", len(tag_counter))

# -----------------------------------------
# First 5000 most frequent tags
# -----------------------------------------

top5000 = tag_counter.most_common(5000)

top5000_tags = set(
    tag for tag, frequency in top5000
)

print("Number of Top 5000 Tags:", len(top5000_tags))

# -----------------------------------------
# Count questions covered by top 5000
# -----------------------------------------

covered_questions = 0

for tags in questions["tags"].fillna(""):
    
    question_tags = set(
        re.findall(
            r"<([^>]+)>",
            str(tags)
        )
    )
    
    if question_tags.intersection(top5000_tags):
        covered_questions += 1

# -----------------------------------------
# Calculate percentage
# -----------------------------------------

percentage = (
    covered_questions / len(questions)
) * 100

print("\n========== Q10 RESULT ==========")

print("Total Questions:", len(questions))

print(
    "Questions Covered by Top 5000 Tags:",
    covered_questions
)

print(
    "Percentage of Questions Covered by First 5000 Tags:",
    round(percentage, 2),
    "%"
)
```
